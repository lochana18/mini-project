```python
def add_expense():
    file = open("expense.txt", "a")
    date = input("Enter Date : ")
    item = input("Enter Item Name : ")
    amount = input("Enter Amount : ")
    file.write(date + "," + item + "," + amount + "\n")
    file.close()
    print("Expense Saved")
def view_expense():
    file = open("expense.txt", "r")
    print("\nDate\tItem\tAmount")
    for line in file:
        print(line.strip())
    file.close()
def total_expense():
    file = open("expense.txt", "r")
    total = 0
    for line in file:
        data = line.strip().split(",")
        total = total + int(data[2])
    file.close()
    print("Total Expense =", total)
for i in range(3):
    print("\n1. Add Expense")
    print("2. View Expense")
    print("3. Total Expense")
    choice = input("Enter Choice : ")
    if choice == "1":
        add_expense()
    if choice == "2":
        view_expense()
    if choice == "3":
        total_expense()
```
